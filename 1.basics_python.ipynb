## Enhanced for loops

```python
# 1. enumerate() — get index + value together
fruits = ["apple","banana","mango"]

for index, fruit in enumerate(fruits):
    print(index,fruit)
    #Without enumerate, people write clunky counters manually (i = 0; i += 1 inside the loop). This replaces that.
```

```python
# 2. zip() = loop over multiple lists in parllel

names = ["siva","raj","kumar"]
ages = [25,30,28]

for name,age in zip(names,ages):
    print(name,age)

#Stops at the shortest list if lengths differ.
```

```python
#3. Looping over a dictionary
person = {"name": "siva", "age": 25, "city": "london"}
for key in person:
    print(key)
print("=================")
for key,value in person.items():
    print(key,value)
print("=================")
for values in person.values():
    print(values)
```

```python
#4. Nested loops
for i in range(3):
    for j in range(3):
        print(i,j)
```

```python
#for...else
#Unusual but real — the else runs only if the loop
#  completes without hitting a break:

for i in range(5):
    if i ==10:
        break
else:
    print("loop finished withput break")
```

```python
# Nested/Multiple comprehensions
matrix = [[i*j for j in range(3)] for i in range(3)]
print(matrix)
```

```python
#7. Comprehension with condition (filter)
evens = [x for x in range(20) if x % 2 == 0]
print(evens)
```

## Functions

```python
#
# Basic definition
def greet(name):
    return f"Hello, {name}"

greet("Siva")
```

```python
#2. Default arguments
def greet(name,greeting="Hello"):
    return f"{greeting},{name}"

greet("Siva")
greet("Siva","Hey")
```

```python
#3. Positional vs keyword arguments
def describe(name,age):
    print(f"{name}is {age}")

describe("Siva ",25) #positional

describe(age=25,name="Siva ")#here keyword order doesnt matter here
```

```python
#*args — variable number of positional arguments
def all_all(*numbers):
    return sum(numbers)

all_all(1,2,3,4)
#*args constantly in library code — it lets a function accept any number 
# of inputs without you predefining how many.
```

```python
#**kwargs — variable number of keyword arguments
def build_profile(**details):
    return details

build_profile(name="Siva", age=25, city="London")
# details becomes the dict {'name': 'Siva', 'age': 25, 'city': 'London'}
#This is exactly how frameworks like LangGraph or scikit-learn let you pass 
# arbitrary config options into a constructor —SomeClass(**kwargs) forwards 
# whatever you give it.
```

```python
#Combining all argument types

def full_example(a,b,*args,c=10,**kwargs):
    print(a,b,args,c,kwargs)

#Order matters: normal args → *args → keyword-only defaults → **kwargs
```

```python
# Scope — local, global, nonlocal

x = "global"

def outer():
    y = "enclosing"
    def inner():
        nonlocal y
        y = "changed"
    inner()
    print(y)          # "changed"

def modify_global():
    global x
    x = "modified"
```
